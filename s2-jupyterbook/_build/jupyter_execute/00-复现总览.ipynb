这本书把原文的 11 张机理图、6 张表逐项重算，并把**原文没有说清的两处**单独拆开：

1. **变尺度里的 $\nu$ 到底是什么**（第 3 章）。原文只写"调 $R$，相变那个 $R$ 就是 $\omega$"，
   没说代码里怎么实现、频率未知时怎么设。
2. **Poincaré 判据的采样方式**（第 2 章）。若按"每 $N$ 步采一次"实现，
   $h$ 不整除 $2\pi$ 时会造出 0.1 量级的伪离散度，把周期态误判成混沌。
